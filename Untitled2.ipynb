```python
import glob, re
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

R_CH, R_LOAD = 200.0, 426.0

def fit_exp(t_seg, v_seg, p0):
    t0 = t_seg[0]
    f = lambda tt, v_final, v_start, tau: v_final + (v_start - v_final) * np.exp(-(tt - t0) / tau)
    popt, _ = curve_fit(f, t_seg, v_seg, p0=p0)
    return popt, f
```

```python
N_EXPECTED = 240
cols = ["t_ms", "gate", "vcap", "vdrain", "i_mA"]

pulses, current = [], None
for line in open("raw_log.txt", encoding="utf-8", errors="ignore").read().splitlines():
    line = re.sub(r"^\d{1,2}:\d{2}:\d{2}\.\d+\s*->\s*", "", line.strip())
    if line == "START":
        current = []
    elif line == "END":
        if current is not None:
            pulses.append(current)
        current = None
    elif current is not None:
        parts = line.split(",")
        if len(parts) == 5:
            try:
                current.append([float(x) for x in parts])
            except ValueError:
                pass

k = 0
for rows in pulses:
    if len(rows) != N_EXPECTED:
        print(f"skipped a block with {len(rows)} samples (expected {N_EXPECTED})")
        continue
    k += 1
    pd.DataFrame(rows, columns=cols).to_csv(f"pulse{k}.csv", index=False)
    print(f"pulse{k}.csv saved, {len(rows)} samples")
print(f"{k} good pulses found")
```

```python
import serial
import time
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit
import io

# --- CONFIGURATION ---
PORT = 'COM3'  # CHANGE THIS to your Arduino's port (e.g., '/dev/ttyUSB0' on Linux/Mac)
BAUD = 115200
R_CH, R_LOAD = 200.0, 426.0
TARGET_ENERGY_MJ = 5.0  # Target energy in milliJoules for a "good" shot

# --- AI / SMART LOGIC CLASS ---
class SmartPulseController:
    def __init__(self, port, baud):
        self.ser = serial.Serial(port, baud, timeout=2)
        time.sleep(2) # Wait for Arduino reset
        self.history = []
        
    def read_until(self, marker):
        """Helper to read serial until a specific string"""
        data = bytearray()
        while not data.decode('utf-8', errors='ignore').strip().endswith(marker):
            chunk = self.ser.read(1)
            if not chunk:
                break
            data += chunk
        return data.decode('utf-8', errors='ignore').strip()

    def fire_and_analyze(self):
        print("🧠 AI Controller: Requesting system status...")
        self.ser.write(b"STATUS\n")
        status = self.read_until("V_CAP:")
        v_cap = float(status.split("V_CAP:")[1].strip())
        print(f"🔋 Current Capacitor Voltage: {v_cap:.2f} V")

        # Smart Decision 1: Energy Check
        # (Assuming C ≈ 575uF from previous fits)
        C_est = 575e-6 
        energy_mj = 0.5 * C_est * (v_cap ** 2) * 1000
        
        if energy_mj < TARGET_ENERGY_MJ:
            print(f"⚠️ AI DECISION: ABORT. Energy ({energy_mj:.1f} mJ) below target ({TARGET_ENERGY_MJ} mJ). Waiting...")
            return None

        print(f"✅ AI DECISION: Energy sufficient ({energy_mj:.1f} mJ). Initiating FIRE sequence...")
        self.ser.write(b"FIRE\n")
        
        # Wait for data stream
        print("⏳ Capturing transient data...")
        raw_data = self.read_until("END")
        
        # Parse CSV data
        lines = raw_data.split('\n')
        clean_data = [line for line in lines if line.startswith('t_ms') or (line and line[0].isdigit())]
        df = pd.read_csv(io.StringIO('\n'.join(clean_data)))
        
        # --- AI DECISION 2: Anomaly Detection via Curve Fitting ---
        on_indices = np.where(df['gate'] == 1)[0]
        if len(on_indices) > 10:
            i_on, i_off = on_indices[0], on_indices[-1] + 1
            
            # Fit discharge
            t_dis = df['t_ms'].values[i_on:i_off]
            v_dis = df['vcap'].values[i_on:i_off]
            v_final_theory = 5.09 * (R_LOAD / (R_CH + R_LOAD))
            
            try:
                f = lambda tt, v_start, tau: v_final_theory + (v_start - v_final_theory) * np.exp(-(tt - t_dis[0]) / tau)
                popt, _ = curve_fit(f, t_dis, v_dis, p0=[v_dis[0], 80])
                tau_dis = popt[1]
                
                # Estimate R_ds(on) from the initial voltage drop
                vdrain_on = df['vdrain'].values[i_on:i_on+5].mean()
                i_initial = (df['vcap'].values[i_on] - vdrain_on) / R_LOAD * 1000.0 # mA
                r_ds_on = vdrain_on / (i_initial / 1000.0) if i_initial > 0 else 0
                
                # Smart Health Assessment
                health_status = "HEALTHY"
                if r_ds_on > 60.0:
                    health_status = "⚠️ WARNING: MOSFET OVERHEAT / HIGH RESISTANCE"
                if tau_dis > 100.0:
                    health_status = "⚠️ WARNING: CAPACITOR DEGRADATION DETECTED"
                    
                print(f"\n📊 AI ANALYSIS COMPLETE:")
                print(f"   - Tau Discharge: {tau_dis:.1f} ms")
                print(f"   - Estimated R_ds(on): {r_ds_on:.1f} Ω")
                print(f"   - System Health: {health_status}")
                
                # Store for trending
                self.history.append({'tau_dis': tau_dis, 'r_ds_on': r_ds_on, 'energy_mj': energy_mj})
                return df
                
            except RuntimeError:
                print("❌ AI ERROR: Curve fit failed. Noisy data.")
                return None
        return None

    def close(self):
        self.ser.close()

# --- EXECUTE THE SMART LOOP ---
controller = SmartPulseController(PORT, BAUD)

try:
    # Run 3 automated smart cycles
    for i in range(3):
        print(f"\n--- CYCLE {i+1} ---")
        df = controller.fire_and_analyze()
        if df is not None:
            plt.plot(df['t_ms'], df['vcap'], label=f'Cycle {i+1}')
        time.sleep(2) # Wait for cap to recharge
        
    plt.xlabel("Time (ms)")
    plt.ylabel("Capacitor Voltage (V)")
    plt.title("AI-Controlled Adaptive Pulse Firing")
    plt.legend()
    plt.grid(True)
    plt.show()
    
finally:
    controller.close()
```
