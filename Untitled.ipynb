```python
import glob
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

R_CH, R_LOAD, C, VS = 200.0, 470.0, 470e-6, 5.09

def fit_exp(t_seg, v_seg, p0):
    t0 = t_seg[0]
    f = lambda tt, v_final, v_start, tau: v_final + (v_start - v_final) * np.exp(-(tt - t0) / tau)
    popt, _ = curve_fit(f, t_seg, v_seg, p0=p0)
    return popt, f
```

```python
FILE = "pulse1.csv"
df = pd.read_csv(FILE)

fig, ax = plt.subplots(2, 1, sharex=True, figsize=(9, 6))
ax[0].plot(df.t_ms, df.vcap, label="Vcap")
ax[0].plot(df.t_ms, df.vdrain, label="Vdrain")
ax[0].set_ylabel("Voltage (V)"); ax[0].legend(); ax[0].grid(True)
ax[1].plot(df.t_ms, df.i_mA)
ax[1].set_ylabel("Load current (mA)"); ax[1].set_xlabel("Time (ms)"); ax[1].grid(True)
plt.show()
```

```python
import re
import pandas as pd

N_EXPECTED = 240                      # samples per pulse (N in the sketch)
cols = ["t_ms", "gate", "vcap", "vdrain", "i_mA"]

pulses, current = [], None
for line in open("raw_log.txt", encoding="utf-8", errors="ignore").read().splitlines():
    line = re.sub(r"^\d{1,2}:\d{2}:\d{2}\.\d+\s*->\s*", "", line.strip())   # remove timestamps
    if line == "START":
        current = []                  # a new block begins
    elif line == "END":
        if current is not None:
            pulses.append(current)
        current = None
    elif current is not None:
        parts = line.split(",")
        if len(parts) == 5:
            try:
                current.append([float(x) for x in parts])
            except ValueError:
                pass                  # the header line or a garbled line

k = 0
for rows in pulses:
    if len(rows) != N_EXPECTED:
        print(f"skipped a block with {len(rows)} samples (expected {N_EXPECTED})")
        continue
    k += 1
    pd.DataFrame(rows, columns=cols).to_csv(f"pulse{k}.csv", index=False)
    print(f"pulse{k}.csv saved, {len(rows)} samples")
print(f"{k} good pulses found")
```

```python
FILE = "pulse1.csv"
df = pd.read_csv(FILE)
t, v, g = df.t_ms.values, df.vcap.values, df.gate.values
on = np.where(g == 1)[0]
i_on, i_off = on[0], on[-1] + 1

(vf1, vs1, tau1), f1 = fit_exp(t[i_on:i_off], v[i_on:i_off], [3.5, v[i_on], 60])
(vf2, vs2, tau2), f2 = fit_exp(t[i_off:],     v[i_off:],     [5.0, v[i_off], 90])

tau_dis_th = (R_CH * R_LOAD / (R_CH + R_LOAD)) * C * 1e3
tau_rec_th = R_CH * C * 1e3
vfin_th = VS * R_LOAD / (R_CH + R_LOAD)

print(f"Discharge: tau = {tau1:.1f} ms (theory {tau_dis_th:.1f}),  final = {vf1:.2f} V (theory {vfin_th:.2f})")
print(f"Recharge : tau = {tau2:.1f} ms (theory {tau_rec_th:.1f}),  final = {vf2:.2f} V (theory {VS:.2f})")
print(f"Capacitance estimate = {tau2 * 1e-3 / R_CH * 1e6:.0f} uF")
print(f"Peak current = {df.i_mA.max():.1f} mA")

plt.figure(figsize=(8, 4))
plt.plot(t, v, ".", label="measured")
plt.plot(t[i_on:i_off], f1(t[i_on:i_off], vf1, vs1, tau1), label=f"discharge fit, tau = {tau1:.0f} ms")
plt.plot(t[i_off:],     f2(t[i_off:],     vf2, vs2, tau2), label=f"recharge fit, tau = {tau2:.0f} ms")
plt.xlabel("Time (ms)"); plt.ylabel("Vcap (V)"); plt.legend(); plt.grid(True)
plt.show()
```

```python
rows = []
for name in sorted(glob.glob("pulse*.csv")):
    d = pd.read_csv(name)
    t, v, g = d.t_ms.values, d.vcap.values, d.gate.values
    on = np.where(g == 1)[0]
    i_on, i_off = on[0], on[-1] + 1
    try:
        (vf1, vs1, tau1), _ = fit_exp(t[i_on:i_off], v[i_on:i_off], [3.5, v[i_on], 60])
        (vf2, vs2, tau2), _ = fit_exp(t[i_off:],     v[i_off:],     [5.0, v[i_off], 90])
    except RuntimeError:
        print(f"{name}: the fit failed, check this pulse")
        continue
    rows.append({"file": name, "peak_mA": d.i_mA.max(), "v_before": v[:on[0]].mean(),
                 "v_end_on": vf1, "tau_dis_ms": tau1, "tau_rec_ms": tau2,
                 "C_est_uF": tau2 * 1e-3 / R_CH * 1e6})
summary = pd.DataFrame(rows).round(2)
print(summary.to_string(index=False))

for name in sorted(glob.glob("pulse*.csv")):
    d = pd.read_csv(name)
    plt.plot(d.t_ms, d.vcap, label=name)
plt.xlabel("Time (ms)"); plt.ylabel("Vcap (V)"); plt.legend(); plt.grid(True)
plt.show()
```
