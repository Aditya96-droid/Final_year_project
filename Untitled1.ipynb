```python
import pandas as pd
import matplotlib.pyplot as plt
import io

# 1. Read the raw text file
with open('raw_log.txt', 'r') as file:
    lines = file.readlines()

# 2. Clean the data: Keep only lines that contain a comma (the CSV data)
clean_lines = [line for line in lines if ',' in line]

# 3. Combine clean lines into a single string and read as CSV
csv_data = "".join(clean_lines)
df = pd.read_csv(io.StringIO(csv_data))

# 4. Plot the Transient Curve
plt.figure(figsize=(10, 6))

plt.plot(df['time_ms'], df['vcap'], label='Capacitor Voltage (A0)', color='blue', linewidth=2)
plt.plot(df['time_ms'], df['vdrain'], label='MOSFET Drain (A1)', color='red', linestyle='--', linewidth=2)
plt.plot(df['time_ms'], df['vgate'], label='MOSFET Gate (A2)', color='green', alpha=0.7)

# 5. Format for Project Report
plt.title('Capacitor Discharge Transient Response (470µF / 426Ω Load)')
plt.xlabel('Time (milliseconds)')
plt.ylabel('Voltage (V)')
plt.grid(True, alpha=0.3)
plt.legend(loc='upper right')
plt.axhline(y=0, color='black', linewidth=0.5)

plt.show()

# 6. Print a Quick Summary for your Report
print("=== TRANSIENT ANALYSIS SUMMARY ===")
print(f"Starting Capacitor Voltage: {df['vcap'].iloc[0]:.3f} V")
print(f"Ending Capacitor Voltage (at 1500ms): {df['vcap'].iloc[-1]:.3f} V")
print(f"Total Voltage Drop: {df['vcap'].iloc[0] - df['vcap'].iloc[-1]:.3f} V")
print(f"Average MOSFET Drain Voltage (while ON): {df['vdrain'].mean():.3f} V")
print("==================================")
```

```python
import re
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

R_CH, R_LOAD = 200.0, 426.0     # your measured resistor values
```

```python
rows = []
for line in open("capture1.txt", encoding="utf-8", errors="ignore"):
    line = re.sub(r"^\d{1,2}:\d{2}:\d{2}\.\d+\s*->\s*", "", line.strip())   # remove timestamps if present
    parts = line.split(",")
    if len(parts) == 4:
        try:
            rows.append([float(x) for x in parts])
        except ValueError:
            pass                                                           # skip header and text lines
df = pd.DataFrame(rows, columns=["t_ms", "vcap", "vdrain", "vgate"])
print(len(df), "samples,", df.t_ms.max(), "ms")

fig, ax = plt.subplots(3, 1, sharex=True, figsize=(9, 8))
ax[0].plot(df.t_ms, df.vcap); ax[0].set_ylabel("Vcap (V)")
ax[1].plot(df.t_ms, df.vdrain, color="C1"); ax[1].set_ylabel("Vdrain (V)")
ax[2].plot(df.t_ms, df.vgate, color="C2"); ax[2].set_ylabel("Vgate (V)"); ax[2].set_xlabel("Time (ms)")
for a in ax: a.grid(True)
plt.show()
```

```python
d = df[df.t_ms >= 2]                       # skip the first sample, the switch is still turning on
t0 = d.t_ms.iloc[0]
f = lambda tt, vf, vs, tau: vf + (vs - vf) * np.exp(-(tt - t0) / tau)
(vf, vs, tau), _ = curve_fit(f, d.t_ms, d.vcap, p0=[3.5, d.vcap.iloc[0], 80])

v_sup   = df.vcap.iloc[0]
r_par   = R_CH * R_LOAD / (R_CH + R_LOAD)
settled = df[df.t_ms > 1000]
vd      = settled.vdrain.median()

print(f"Supply (Vcap before switching): {v_sup:.3f} V")
print(f"Vcap settles at               : {vf:.3f} V   (divider predicts {v_sup * R_LOAD / (R_CH + R_LOAD):.3f} V)")
print(f"Discharge tau                 : {tau:.1f} ms")
print(f"Capacitance estimate          : {tau * 1e-3 / r_par * 1e6:.0f} uF")
print(f"Vgate while on                : {settled.vgate.mean():.3f} V")
print(f"Vdrain while on (median)      : {vd:.3f} V")

i_in   = (v_sup - vf) / R_CH * 1000
i_load = (vf - vd) / R_LOAD * 1000
i_zero = vf / R_LOAD * 1000
print(f"Current through charge resistor    : {i_in:.2f} mA")
print(f"Current through load, using Vdrain : {i_load:.2f} mA")
print(f"Current through load, Vdrain = 0   : {i_zero:.2f} mA")

late   = df[df.t_ms > 100]
base   = late.vdrain.rolling(41, center=True, min_periods=1).median()
spikes = late[late.vdrain > base + 0.10]
print(len(spikes), "Vdrain samples more than 0.10 V above their local baseline")
```

```python
fig, ax = plt.subplots(2, 1, sharex=True, figsize=(9, 6))
ax[0].plot(d.t_ms, d.vcap, ".", label="measured")
ax[0].plot(d.t_ms, f(d.t_ms, vf, vs, tau), label=f"fit, tau = {tau:.0f} ms")
ax[0].set_ylabel("Vcap (V)"); ax[0].legend(); ax[0].grid(True)
ax[1].plot(d.t_ms, d.vcap - f(d.t_ms, vf, vs, tau))
ax[1].set_ylabel("Measured minus fit (V)"); ax[1].set_xlabel("Time (ms)"); ax[1].grid(True)
plt.show()
```
